```python
def merge_dicts_with_conflict_check(*dicts):
    merged = {}
    conflicts = {}
    
    for d in dicts:
        for key, value in d.items():
            if key in merged:
                # 기존 값과 새 값이 다르면 충돌로 기록
                if merged[key] != value:
                    if key not in conflicts:
                        conflicts[key] = [merged[key]]
                    conflicts[key].append(value)
                    # 값이 같으면 그냥 넘어감 (아무것도 안 함)
            else:
                # 키가 없으면 추가
                merged[key] = value
    
    if conflicts:
        print("⚠️ 키 충돌 발견 (값이 다른 경우):")
        for key, values in conflicts.items():
            print(f"  '{key}': {values}")
    else:
        print("✅ 키 충돌 없음 (또는 모든 충돌 키의 값이 동일함)")
    
    return merged

# 사용
combined = merge_dicts_with_conflict_check(
    fleurs_code_to_name,
    cv_code_to_name,
    omnilingual_code_to_name
)
```

```python
sorted_dict = dict(sorted(combined.items()))
```
